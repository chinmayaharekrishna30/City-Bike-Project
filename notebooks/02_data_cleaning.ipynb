```python
# Normalize station ID columns
trips["start_station_id"] = trips["start_station_id"].astype("string").str.strip()
trips["end_station_id"] = trips["end_station_id"].astype("string").str.strip()
stations["station_id"] = stations["station_id"].astype("string").str.strip()

# Remove missing station IDs from the station lookup table
stations = stations.dropna(subset=["station_id"]).drop_duplicates(
    subset=["station_id"]
)

# Rename station columns
start_stations = stations.rename(columns={
    "station_id": "start_station_id",
    "station_name": "start_station_info_name",
    "latitude": "start_station_info_lat",
    "longitude": "start_station_info_lng"
})

end_stations = stations.rename(columns={
    "station_id": "end_station_id",
    "station_name": "end_station_info_name",
    "latitude": "end_station_info_lat",
    "longitude": "end_station_info_lng"
})

# Merge
merged = (
    trips
    .merge(start_stations, on="start_station_id", how="left")
    .merge(end_stations, on="end_station_id", how="left")
)

# Check unmatched stations
print("Unmatched start stations:",
      merged["start_station_info_name"].isna().sum())
print("Unmatched end stations:",
      merged["end_station_info_name"].isna().sum())

merged.head()
```

```python
import pandas as pd

trips = pd.read_csv(
    r"C:\City Bike\data\processed\citibike_cleaned.csv",
    dtype={"start_station_id": "string", "end_station_id": "string"}
)

stations = pd.read_csv(
    r"C:\City Bike\notebooks\cleaned_data.csv",
    dtype={"station_id": "string"}
)

# Standardize station IDs
def clean_id(series):
    return (
        series.astype("string")
        .str.strip()
        .str.replace(r"\.0$", "", regex=True)
    )

trips["start_station_id"] = clean_id(trips["start_station_id"])
trips["end_station_id"] = clean_id(trips["end_station_id"])
stations["station_id"] = clean_id(stations["station_id"])

stations = stations.dropna(subset=["station_id"]).drop_duplicates("station_id")

start_stations = stations.rename(columns={
    "station_id": "start_station_id",
    "station_name": "start_station_info_name",
    "latitude": "start_station_info_lat",
    "longitude": "start_station_info_lng"
})

end_stations = stations.rename(columns={
    "station_id": "end_station_id",
    "station_name": "end_station_info_name",
    "latitude": "end_station_info_lat",
    "longitude": "end_station_info_lng"
})

merged = (
    trips
    .merge(start_stations, on="start_station_id", how="left")
    .merge(end_stations, on="end_station_id", how="left")
)

# Keep rows and columns; replace missing values
merged = merged.fillna("Unknown")

output_path = r"C:\City Bike\data\processed\citibike_trips_with_stations.csv"
merged.to_csv(output_path, index=False)

print("Rows and columns:", merged.shape)
print("Remaining NaN values:", merged.isna().sum().sum())
print("Saved:", output_path)
```
